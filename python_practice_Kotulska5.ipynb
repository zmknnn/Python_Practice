6. (11б). Створіть 2 дочірніх класи від 3d-фігура: Циліндр та Куб, які мають атрибути, необхідні для розрахунку об'єму. Визначте метод get_volume в кожному з них так, щоби вони розраховували об'єм у залежності від формули для кожного типу фігури. Створіть обʼєкт класу Циліндр та обʼєкт класу Куб. Виконайте операцію суми за допомогою оператора + між ними, аби знайти сумарний об'єм.

```python
import math

class Figure3D:
    def get_volume(self):
        return 0

    def __add__(self, other):
        return self.get_volume() + other.get_volume()

class Cylinder(Figure3D):
    def __init__(self, radius, height):
        self.radius = radius
        self.height = height

    def get_volume(self):
        return math.pi * self.radius ** 2 * self.height

class Cube(Figure3D):
    def __init__(self, side):
        self.side = side

    def get_volume(self):
        return self.side ** 3

cyl = Cylinder(2, 7)
cube = Cube(3)

total_volume = cyl + cube
print(total_volume)
```

8. (13б). Створіть клас EWallet з приватними атрибутами balance та id. Реалізуйте методи поповнення та зняття коштів, забезпечивши належну інкапсуляцію. Підказка: використовуйте декоратори property та setter.

```python
class EWallet:
    def __init__(self, wallet_id, balance=0.0):
        self.__id = wallet_id
        self.__balance = float(balance)

    @property
    def id(self):
        return self.__id

    @property
    def balance(self):
        return self.__balance

    @balance.setter
    def balance(self, amount):
        if amount < 0:
            raise ValueError("Баланс не може бути від'ємним")
        self.__balance = float(amount)

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Сума поповнення має бути більшою за 0")
        self.balance = self.__balance + amount

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Сума зняття має бути більшою за 0")
        if amount > self.__balance:
            raise ValueError("Недостатньо коштів")
        self.balance = self.__balance - amount


w = EWallet("Nonna", 700)
w.deposit(320)
w.withdraw(590)
print(w.id, w.balance)
```

9. (11б). Створіть клас ShoppingCart, який містить список об'єктів типу Item. Реалізуйте методи для додавання айтемів, їх видалення та відображення списку айтемів.

```python
class Item:
    def __init__(self, name, price):
        self.name = name
        self.price = float(price)

    def __str__(self):
        return f"{self.name}: {self.price} грн"


class ShoppingCart:
    def __init__(self):
        self.items = []

    def add_item(self, item: Item):
        self.items.append(item)

    def remove_item(self, item: Item):
        if item in self.items:
            self.items.remove(item)

    def show_items(self, title=None):
        if title:
            print(f"--- {title} ---")
        if not self.items:
            print("Кошик порожній")
        else:
            for item in self.items:
                print(item)


cart = ShoppingCart()
item1 = Item("Хлібчик", 30)
item2 = Item("Чоколад", 100)
item3 = Item("Кукумбер", 7)

cart.add_item(item1)
cart.add_item(item2)
cart.add_item(item3)

cart.show_items("До видалення")

cart.remove_item(item2)

cart.show_items("Після видалення")
```

13. (11б). Створіть клас, що має визначений Iterator Protocol та при кожному виклику next повертає назву дня тижня, поки вони не вичерпаються. Зверніть увагу, що порядок має значення.

```python
class DaysOfWeekIterator:
    def __init__(self):
        self.days = ["Monday", "Tuesday", "Wednesday",
                     "Thursday", "Friday", "Saturday", "Sunday"]
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index >= len(self.days):
            raise StopIteration

        day = self.days[self.index]
        self.index += 1
        return day

print("All days of the week:")
for day in DaysOfWeekIterator():
    print(day)
```
