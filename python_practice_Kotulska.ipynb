15. Порівняйте список та кортеж. Назвіть схожості та відмінності, випадки використання.

**Схожість:**
Кортеж, як і список, є впорядкованою послідовністю елементів. Також кортеж і список може містити різні типи елементів та містити інші кортежі та списки.  
**Відмінність:**
Різниця між ними та, що кортежі відносяться до незмінних типів даних. Тобто, кортеж - це список, елементи якого не можна змінити. Також під час створення елементи списку вказуються у квадратних дужках [], а елементи кортежу — у круглих ().
**Випадки використання:**

**Список:** змінювані дані. Додавання, видалення, редагування елементів.

**Кортеж:** незмінні дані. Константи, координати, ключі словника

19. Створіть дві множини. Продемонстуйте над ними операції: обʼєднання, різниці, пересічі та симетричної різниці. Використовуйте методи, що не змінюють множини, а створюють нові.

````python
A = {1, 3, 5}
B = {0, 2, 4, 5, 9, 1}
print('Обʼєднання:', A | B)
print('Різниця:', A - B)
print('Пересіч:', A & B)
print('Симетричної різниця:', A ^ B)
````
